```python
from evaluation.Metrics import Metrics
from evaluation.Visualizer import Visualizer
from pathlib import Path 
import pandas as pd
```

```python
metrics = []
for csv_path in Path("../results").glob("*.csv"):
    metric = Metrics(csv_path)
    metrics.append(metric)

    out_metrics_dir = Path("../results/metrics")
    out_metrics_dir.mkdir(parents=True, exist_ok=True)
    out_metrics_path = out_metrics_dir / f"{csv_path.stem}.csv"
    metric.metrics.to_csv(out_metrics_path)
   
    output_plot_dir = Path("../results/plots")
    output_plot_dir.mkdir(parents=True, exist_ok=True)
    output_plot_path = output_plot_dir / f"{csv_path.stem}.png"
    Visualizer.print_metrics_summary(metric.metrics, csv_path.stem, save_path=output_plot_path)
```

```python
# Grok vs Gpt Oss
output_plot_dir = Path("../results/plots/aggregated")
output_plot_dir.mkdir(parents=True, exist_ok=True)

# Data Prep
grok_overall, grok_perepisode = Metrics.aggregate([m for m in metrics if "grok" in m.metadata.get('filename')], name="Grok4.1Fast: Overall")
gpt_overall, gpt_perepisode = Metrics.aggregate([m for m in metrics if "gpt" in m.metadata.get('filename')], name="GptOss: Overall")

# Overall
output_plot_path = output_plot_dir / f"GrokFast_vs_GptOss_Overall.png"
Visualizer.print_aggregated_metrics_summary([grok_overall, gpt_overall], "grok-4.1-fast vs gpt-oss-120b", save_path=output_plot_path)

# Per Episode
output_plot_path = output_plot_dir / f"GrokFast_vs_GptOss_PerEpisode.png"
Visualizer.print_aggregated_metrics_perepisode([grok_perepisode, gpt_perepisode], "grok-4.1-fast vs gpt-oss-120b", save_path=output_plot_path)
```
