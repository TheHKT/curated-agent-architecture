```python
from evaluation.Metrics import Metrics
from evaluation.Visualizer import Visualizer
from pathlib import Path 
import pandas as pd
```

```python
metrics = []
for csv_path in Path("../results").glob("*.csv"):
    metric = Metrics(csv_path)
    metrics.append(metric)

    out_metrics_dir = Path("../results/metrics")
    out_metrics_dir.mkdir(parents=True, exist_ok=True)
    out_metrics_path = out_metrics_dir / f"{csv_path.stem}.csv"
    metric.metrics.to_csv(out_metrics_path)
   
    output_plot_dir = Path("../results/plots")
    output_plot_dir.mkdir(parents=True, exist_ok=True)
    output_plot_path = output_plot_dir / f"{csv_path.stem}.png"
    Visualizer.print_metrics_summary(metric.metrics, csv_path.stem, save_path=output_plot_path)
```

## ShadowEnv=copy() VS ShadowEnv=llm

```python
output_plot_dir = Path("../results/plots/aggregated")
output_plot_dir.mkdir(parents=True, exist_ok=True)

# Overall
llmaction_true_metrics_overall = Metrics.aggregate_metrics([m for m in metrics if m.metadata.get('llmaction') == 'True'], name="LLM")
llmaction_false_metrics_overall = Metrics.aggregate_metrics([m for m in metrics if m.metadata.get('llmaction') == 'False'], name="COPY")
output_plot_path = output_plot_dir / f"ShadowEnv_Overall.png"
Visualizer.print_aggregated_metrics_summary([llmaction_true_metrics_overall, llmaction_false_metrics_overall], "ShadowEnv: Overall", save_path=output_plot_path)

llmaction_true_metrics_overall_perepisode = Metrics.aggregate_metrics_perepisode([m for m in metrics if m.metadata.get('llmaction') == 'True'], name="LLM")
llmaction_false_metrics_overall_perepisode = Metrics.aggregate_metrics_perepisode([m for m in metrics if m.metadata.get('llmaction') == 'False'], name="COPY")
output_plot_path = output_plot_dir / f"ShadowEnv_Overall_PerEpisode.png"
Visualizer.print_aggregated_metrics_perepisode([llmaction_true_metrics_overall_perepisode, llmaction_false_metrics_overall_perepisode], "ShadowEnv: Overall Per Episode", save_path=output_plot_path.with_name(f"ShadowEnv_Overall_PerEpisode.png"))

# With Lookahead
llmaction_true_metrics_look = Metrics.aggregate_metrics([m for m in metrics if m.metadata.get('llmaction') == 'True' and int(m.metadata.get('lookaheadsize')) > 1], name="LLM")
llmaction_false_metrics_look = Metrics.aggregate_metrics([m for m in metrics if m.metadata.get('llmaction') == 'False' and int(m.metadata.get('lookaheadsize')) > 1], name="COPY")
output_plot_path = output_plot_dir / f"ShadowEnv_Lookahead.png"
Visualizer.print_aggregated_metrics_summary([llmaction_true_metrics_look, llmaction_false_metrics_look], "ShadowEnv: Lookahead", save_path=output_plot_path)

# With no Lookahead
llmaction_true_metrics_nolook = Metrics.aggregate_metrics([m for m in metrics if m.metadata.get('llmaction') == 'True' and int(m.metadata.get('lookaheadsize')) == 1], name="LLM") 
llmaction_false_metrics_nolook = Metrics.aggregate_metrics([m for m in metrics if m.metadata.get('llmaction') == 'False' and int(m.metadata.get('lookaheadsize')) == 1], name="COPY")
output_plot_path = output_plot_dir / f"ShadowEnv_NoLookahead.png"
Visualizer.print_aggregated_metrics_summary([llmaction_true_metrics_nolook, llmaction_false_metrics_nolook], "ShadowEnv: No Lookahead", save_path=output_plot_path)
```
